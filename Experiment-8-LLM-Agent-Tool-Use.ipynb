```python
#Task 1 — Import libraries
import os
from langchain_openai import ChatOpenAI
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_core.prompts import ChatPromptTemplate

print("Libraries successfully imported!")
```

```python
#Task 2 — Create the Calculator tool
def calculator(expression: str) -> str:
    try:
        result = eval(expression)
        return str(result)
    except Exception as e:
        return f"Error: {str(e)}"
```

```python
print(calculator("342 * 19"))
```

```python
#Task 3 — Create WebSearch
search = DuckDuckGoSearchRun()
```

```python
search_result = search.run("current Prime Minister of India")
print(search_result)
```

```python
#Task 3 — Wrap tools using Tool
calculator_tool = Tool(
    name="Calculator",
    func=calculator,
    description="Useful for evaluating arithmetic expressions, for example 23*17 or 342*19."
)

web_search_tool = Tool(
    name="WebSearch",
    func=search.run,
    description="Useful for searching the web for current, recent, or factual information that may not be known from the model's memory."
)
```

```python
tools = [
    calculator_tool,
    web_search_tool
]

print("Tools created successfully.")
```

```python
#Task 4 — Create the LLM
llm = ChatOpenAI(
    model="gpt-3.5-turbo",
    temperature=0,
    api_key=os.environ.get("OPENAI_API_KEY")
)

print("LLM created successfully.")
```

```python
#Task 4 — Build the Agent
agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

print("Agent created successfully.")
```

```python
#Task 5 — Run the main question
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import initialize_agent, AgentType
from getpass import getpass
import os

# Securely prompt for the Gemini API Key to avoid metadata loop issues
if not os.environ.get("GOOGLE_API_KEY"):
    os.environ["GOOGLE_API_KEY"] = getpass("Enter your Google Gemini API key: ")

# Initialize the LLM using the free gemini-2.5-flash model with an explicit API key
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0,
    google_api_key=os.environ["GOOGLE_API_KEY"]
)

# Re-initialize the agent with the Gemini LLM
agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

question = "What is 342 * 19, and who is the current Prime Minister of India?"

result = agent.invoke(question)

print("\nFINAL ANSWER:")
print(result["output"])
```

```python
#Task 6 — General knowledge question
result3 = agent.invoke(
    "What is the capital of France?"
)

print("\nFINAL ANSWER:")
print(result3["output"])
```
